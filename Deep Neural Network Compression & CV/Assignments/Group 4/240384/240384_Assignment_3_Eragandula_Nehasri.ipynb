````python
from PIL import Image
import io
import base64
import matplotlib.pyplot as plt
import numpy as np
import cv2
````

````python
img = Image.open("myimage.png")
img = np.array(img)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")
````

````python
def convolve(img, kernel):
    kh, kw = kernel.shape
    pad_h = kh // 2
    pad_w = kw // 2
    # Pad image using reflection
    img_padded = np.pad(img, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')
    output = np.zeros_like(img, dtype=np.float32)
    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            region = img_padded[i:i+kh, j:j+kw]
            output[i, j] = np.sum(region * kernel)
    return output
````

````python
# This function performs 2D convolution using a sliding window and reflection padding to handle image boundaries.
````

````python
# Converting RGB image to grayscale using NumPy
R, G, B = img[:,:,0], img[:,:,1], img[:,:,2]
gray = (0.299*R+0.587*G+0.114*B).astype(np.float32)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")
````

````python
def average_blur_gray(img, ksize):
    kernel = np.ones((ksize, ksize), dtype=np.float32)
    kernel = kernel / (ksize * ksize)
    return convolve(img, kernel)
````

````python
blur_gray = average_blur_gray(gray, ksize=7)
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.imshow(gray, cmap='gray')
plt.title("Original Grayscale")
plt.axis("off")
plt.subplot(1,2,2)
plt.imshow(blur_gray, cmap='gray')
plt.title("Average Blurred (Grayscale)")
plt.axis("off")
plt.show()
````

````python
# Average blur smooths the image by replacing each pixel with the average of its neighboring pixels.
````

````python
def average_blur_rgb(img, ksize):
    img = img.astype(np.float32)
    R_blur = average_blur_gray(img[:,:,0], ksize)
    G_blur = average_blur_gray(img[:,:,1], ksize)
    B_blur = average_blur_gray(img[:,:,2], ksize)
    blurred = np.stack([R_blur, G_blur, B_blur], axis=2)
    blurred = np.clip(blurred, 0, 255)
    return blurred.astype(np.uint8)
````

````python
blur_rgb = average_blur_rgb(img, ksize=7)
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original RGB")
plt.axis("off")
plt.subplot(1,2,2)
plt.imshow(blur_rgb)
plt.title("Average Blurred RGB")
plt.axis("off")
plt.show()
````

````python
def gaussian_kernel(ksize, sigma):
    assert ksize % 2 == 1, "Kernel size must be odd"
    k = ksize // 2
    kernel = np.zeros((ksize, ksize), dtype=np.float32)
    for i in range(ksize):
        for j in range(ksize):
            x = i - k
            y = j - k
            kernel[i, j] = np.exp(-(x**2+y**2) / (2*sigma**2))
    kernel = kernel / np.sum(kernel)
    return kernel
````

````python
# This function generates a normalized 2D Gaussian kernel based on the given kernel size and standard deviation.
````

````python
gauss_k = gaussian_kernel(ksize=7, sigma=1.5)
blur_gauss_gray = convolve(gray, gauss_k)
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.imshow(gray, cmap='gray')
plt.title("Original Grayscale")
plt.axis("off")
plt.subplot(1,2,2)
plt.imshow(blur_gauss_gray, cmap='gray')
plt.title("Gaussian Blurred (Grayscale)")
plt.axis("off")
plt.show()
````

````python
def gaussian_blur_rgb(img, ksize, sigma):
    img = img.astype(np.float32)
    kernel = gaussian_kernel(ksize, sigma)
    R = convolve(img[:,:,0], kernel)
    G = convolve(img[:,:,1], kernel)
    B = convolve(img[:,:,2], kernel)
    blurred = np.stack([R, G, B], axis=2)
    blurred = np.clip(blurred, 0, 255)
    return blurred.astype(np.uint8)
````

````python
blur_gauss_rgb = gaussian_blur_rgb(img, ksize=7, sigma=1.5)
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original RGB")
plt.axis("off")
plt.subplot(1,2,2)
plt.imshow(blur_gauss_rgb)
plt.title("Gaussian Blurred RGB")
plt.axis("off")
plt.show()
````

````python
# Gaussian blur smooths the image by giving more weight to nearby pixels and less to distant ones.
````

#### Task - Sobel Edge Detection (NumPy Only)
Implement Sobel X and Sobel Y filters manually using convolution. Apply on the grayscale image.

Compute:
1. Gx
2. Gy
3. Gradient magnitude

Normalize to 0–255 and save the edge map. Threshold if you wish to, if you do then make the thresholding part clear using code comments.

````python
# Sobel kernels
sobel_x = np.array([[-1, 0, 1],
                    [-2, 0, 2],
                    [-1, 0, 1]], dtype=np.float32)
sobel_y = np.array([[ 1,  2,  1],
                    [ 0,  0,  0],
                    [-1, -2, -1]], dtype=np.float32)
````

````python
Gx = convolve(gray, sobel_x)
Gy = convolve(gray, sobel_y)
````

````python
gradient_magnitude = np.sqrt(Gx**2 + Gy**2)
````

````python
gradient_magnitude = gradient_magnitude / np.max(gradient_magnitude) * 255
gradient_magnitude = gradient_magnitude.astype(np.uint8)
````

````python
plt.figure(figsize=(12,4))
plt.subplot(1,4,1)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")
plt.subplot(1,4,2)
plt.imshow(Gx, cmap='gray')
plt.title("Sobel Gx")
plt.axis("off")
plt.subplot(1,4,3)
plt.imshow(Gy, cmap='gray')
plt.title("Sobel Gy")
plt.axis("off")
plt.subplot(1,4,4)
plt.imshow(gradient_magnitude, cmap='gray')
plt.title("Gradient Magnitude")
plt.axis("off")
plt.show()
````

````python
#thresholding
threshold = 100
edges_thresh = gradient_magnitude > threshold
plt.imshow(edges_thresh, cmap='gray')
plt.title("Thresholded Edges")
plt.axis("off")
````

````python
# Thresholding suppresses weak edges and retains strong edges.
````

````python
# FFT of original image
f_orig = np.fft.fft2(gray)
fshift_orig = np.fft.fftshift(f_orig)
mag_orig = np.log(1 + np.abs(fshift_orig))
````

````python
# FFT of sharpened image
f_sharp = np.fft.fft2(sharp_usm)
fshift_sharp = np.fft.fftshift(f_sharp)
mag_sharp = np.log(1 + np.abs(fshift_sharp))
````

````python
plt.figure(figsize=(15,5))
plt.subplot(1,3,1)
plt.imshow(mag_orig, cmap='gray')
plt.title("Original Image Spectrum")
plt.axis("off")
plt.subplot(1,3,2)
plt.imshow(mag_blur, cmap='gray')
plt.title("Gaussian Blurred Spectrum")
plt.axis("off")
plt.subplot(1,3,3)
plt.imshow(mag_sharp, cmap='gray')
plt.title("Sharpened Image Spectrum")
plt.axis("off")
plt.show()
````

````python
# Observations:
# The original image spectrum shows strong low-frequency components at the center.
# After Gaussian blurring, high-frequency components are suppressed, making the outer regions of the spectrum darker.
# In the sharpened image, high-frequency components are emphasized, resulting in brighter regions away from the center.
# This verifies that convolution kernels act as frequency-domain filters.
````
