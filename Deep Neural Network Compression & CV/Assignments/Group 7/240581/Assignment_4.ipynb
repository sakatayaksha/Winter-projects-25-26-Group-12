To preserve an image in Google Colab without a continuous runtime or using Google Drive, you can use base64 string encoding of images ( its really simple )

First, begin by normally uploading the image into the contents folders as you normally do in colab or fetch it using Drive or any API.

To get the base64 string :
```
import base64
with open("your_image.jpg", "rb") as img_file:
    print(base64.b64encode(img_file.read()).decode())
```

Note that the base64 string is a very long string, click on the three dot icon to the left of the current code cell and click on <code> Copy Cell Output </code> to easily copy it to your clipboard.

Then, to embed the string into your Jupyter Notebook :
```
from PIL import Image
import io
import base64
import matplotlib.pyplot as plt
import numpy as np

# PASTE BASE64 STRING HERE
base64_image = "iVBORw0K..."  # long base64 string

img_data = base64.b64decode(base64_image)
img = Image.open(io.BytesIO(img_data))

plt.imshow(img)
plt.axis("off")
```

````python
sobel_x = np.array([[-1, 0, 1],
                    [-2, 0, 2],
                    [-1, 0, 1]])

sobel_y = np.array([[-1, -2, -1],
                    [ 0,  0,  0],
                    [ 1,  2,  1]])

def convolution(img, kernel):
    h, w = img.shape
    kh, kw = kernel.shape
    pad_h= kh // 2
    pad_w= kw // 2

    pad = np.pad(img, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')
    op = np.zeros_like(img, dtype=float)

    for i in range(h):
        for j in range(w):
            rgn = pad[i:i+kh, j:j+kw]
            op[i, j] = np.sum(rgn * kernel)

    return op
````

````python
def sobel_analysis(gray, threshold=None, title="Image"):
    # Sobel gradients
    Ix = convolution(gray, sobel_x)
    Iy = convolution(gray, sobel_y)

    # Optional thresholding
    if threshold is not None:
        mask = np.sqrt(Ix**2 + Iy**2) > threshold
        Ix_plot = Ix[mask]
        Iy_plot = Iy[mask]

    else:
        Ix_plot = Ix.flatten()
        Iy_plot = Iy.flatten()

    #plotting
    plt.figure(figsize=(12,4))

    plt.subplot(1,3,1)
    plt.imshow(Ix, cmap='gray')
    plt.title(f"{title} Ix")
    plt.axis('off')

    plt.subplot(1,3,2)
    plt.imshow(Iy, cmap='gray')
    plt.title(f"{title} Iy")
    plt.axis('off')

    # Scatter plot
    plt.subplot(1,3,3)
    plt.scatter(Ix_plot, Iy_plot, s=1)
    plt.xlabel("Ix")
    plt.ylabel("Iy")
    plt.title("Iy vs Ix scatter plot")

    plt.tight_layout()
    plt.show()

    return Ix, Iy
````

````python
def harris_r_map(gray_img, window_size=3, k=0.04):

    Ix = convolution(gray_img, sobel_x)               #gradients
    Iy = convolution(gray_img, sobel_y)

    h, w = gray_img.shape
    oft = window_size // 2

    R_map = np.zeros((h, w))


    for i in range(oft, h - oft):               #looping over each pixels
        for j in range(oft, w - oft):


            Ix_w = Ix[i-oft:  i + oft  + 1, j -  oft :j+oft+1]
            Iy_w = Iy[i-oft:  i + oft  + 1, j -oft: j+oft+1]


            a = np.sum(Ix_w ** 2)
            b = np.sum(Ix_w * Iy_w)
            c = np.sum(Iy_w ** 2)


            eg = a + c                            # Eigenvalues
            det = a*c - b*b
            temp = np.sqrt((a - c)**2 + 4*b*b)

            lambda1 = (eg + temp) / 2
            lambda2 = (eg - temp) / 2


            r = det - k * (eg ** 2)       #response

            R_map[i, j] = r

    return R_map
````

````python
import cv2
import matplotlib.pyplot as plt

img=cv2.imread('Horizontal-Lines-in-Photography.jpg')
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.figure(figsize=(15,10))
plt.subplot(1,2,1)
plt.imshow(img)
plt.title('Original')
plt.axis("off")
plt.subplot(1,2,2)
img_gray=cv2.cvtColor(img,cv2.COLOR_RGB2GRAY)
plt.imshow(img_gray,cmap='gray')
plt.title('Grayscale')
plt.axis("off")
plt.show()
````

````python
def hough_slope(img_gray,m_min=-1, m_max=1, m_steps=150, c_min=-300, c_max=300, c_steps=300, canny1=50, canny2=150,threshold_ratio=0.9):


    edges = cv2.Canny(img_gray, canny1, canny2)              #canny edge detection
    plt.figure(figsize=(18,10))
    plt.subplot(1,3,1)
    plt.imshow(edges, cmap='gray')
    plt.title("Canny Edge Map (slope)")
    plt.axis("off")



    m_vals = np.linspace(m_min, m_max, m_steps)           #parameter space
    c_vals = np.linspace(c_min, c_max, c_steps)

    accumulator = np.zeros((m_steps, c_steps))


    ys, xs = np.where(edges > 0)                           #voting

    for x, y in zip(xs, ys):                               #voting for all slopes for every edge pixel
        for i, m in enumerate(m_vals):
            c = y - m * x

            if c < c_min or c > c_max:
                continue

            j = int((c - c_min) / (c_max - c_min) * (c_steps - 1))
            accumulator[i, j] += 1

    #normalised accumulator
    acc_norm = accumulator / (accumulator.max() )
    plt.subplot(1,3,2)

    plt.imshow(acc_norm, cmap='hot')

    plt.title("Accumulator (slope)")


    #for threshold
    indices = np.where(acc_norm > threshold_ratio)

    return edges, accumulator, m_vals, c_vals, indices
````

````python
def line(img_color, m_vals, c_vals, indices):
    img_op = img_color.copy()             #copy for drawing line

    for i, j in zip(indices[0], indices[1]):      #for all possible slopes
        m = m_vals[i]
        c = c_vals[j]

        x1, x2 = 0, img_color.shape[1]
        y1 = int(m * x1 + c)
        y2 = int(m * x2 + c)

        cv2.line(img_op, (x1, y1), (x2, y2), (0, 0, 255), 1)       #draw line

    return img_op
````

````python
#plotting
edges_mc, acc_mc, m_vals, c_vals, idx_mc = hough_slope(img_gray)

img_mc = line(img, m_vals, c_vals, idx_mc)
plt.subplot(1,3,3)
plt.imshow(cv2.cvtColor(img_mc, cv2.COLOR_BGR2RGB))

plt.title("Slope intercpt hough lines")
plt.axis("off")
plt.show()
````

````python
def hough_rho(img_gray, rho_steps=300,theta_steps=180,canny1=50,canny2=150,threshold_ratio=0.9):

    edges = cv2.Canny(img_gray, canny1, canny2)

    plt.figure(figsize=(18,10))                 #canny edge
    plt.subplot(1,3,1)

    plt.imshow(edges, cmap='gray')
    plt.title("Canny Edge Map (rho theta)")
    plt.axis("off")


    h, w = img_gray.shape
    diag = int(np.sqrt(h*h + w*w))         #possible rho

    rhos = np.linspace(-diag, diag, rho_steps)
    thetas = np.linspace(-np.pi/2, np.pi/2, theta_steps)


    accumulator = np.zeros((rho_steps, theta_steps))    #accum for voting

    ys, xs = np.where(edges > 0)                       #voting for all theta in pixel
    for x, y in zip(xs, ys):
        for j, theta in enumerate(thetas):
            rho = x*np.cos(theta) + y*np.sin(theta)
            i = int((rho + diag) / (2*diag) * (rho_steps - 1))
            if 0 <= i < rho_steps:
                accumulator[i, j] += 1

    acc_norm = accumulator / (accumulator.max())
    indices = np.where(acc_norm > threshold_ratio)      #peak eaxtraction using threshold
    plt.subplot(1,3,2)

    plt.imshow(acc_norm, cmap='hot')
    plt.title("Accumulator (rho theta)")



    return edges, accumulator, rhos, thetas, indices
````

````python
def lines_rho(img_color, rhos, thetas, indices):
    img_out = img_color.copy()

    for i, j in zip(indices[0], indices[1]):  #loop over rho and theta
        rho = rhos[i]
        theta = thetas[j]

        a = np.cos(theta)         #polar to cartesian
        b = np.sin(theta)

        x0 = a * rho
        y0 = b * rho

        x1 = int(x0 + 1000 * (-b))
        y1 = int(y0 + 1000 * (a))
        x2 = int(x0 - 1000 * (-b))
        y2 = int(y0 - 1000 * (a))

        cv2.line(img_out, (x1, y1), (x2, y2), (255, 0, 0), 1)      #draw

    return img_out
````

````python
edges_rt, acc_rt, rhos, thetas, idx_rt = hough_rho(img_gray)

img_rt = lines_rho(img, rhos, thetas, idx_rt)
plt.subplot(1,3,3)
plt.imshow(cv2.cvtColor(img_rt, cv2.COLOR_BGR2RGB))
plt.title("Rho Theta Hough lines")
plt.axis("off")
plt.show()
````
